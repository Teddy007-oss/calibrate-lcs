EDA for Clarity Sensor Data

```python
import pandas as pd

df = pd.read_excel(
    r"C:\Users\USER\Downloads\clarity.xlsx",
    parse_dates=["date"],
    index_col="date"
)
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (8, 4))
plt.plot(df.index, df["FEM"], label = "Reference PM2.5", color = "red")
plt.plot(df.index, df["Clarity"], label = "Clarity PM2.5", color = "blue")
plt.xlabel("Date")
plt.ylabel("PM2.5 Concentration (µg/m³)")
plt.title("Clarity Sensor Data (15min (RAW) averaged)")
plt.legend()
plt.show()
```

From the plot above, it is clear that we can use the periods Jan 2024 to March 2024 for Dy period calibration. We can use the period April 2024 to June 2024 for the Wet period calibration.
