Zefan EDA

zefan dry


```python
import pandas as pd

df_dry = pd.read_csv(
    r"C:\Users\USER\Documents\Afri-SET\Zefan\zefan_ref_dry.csv",
    parse_dates=["date"],
    index_col="date"
)   
```

```python
df_dry.describe()
```

```python
data = ["PM2.5", "pm25a", "pm25b", "pm25c"]

dfDry = df_dry[data]

dfDry.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot(dfDry.index, dfDry["PM2.5"], label="Reference PM2.5", color="red")
plt.plot(dfDry.index, dfDry["pm25a"], label="Sensit A", color="green", alpha=0.7)
plt.plot(dfDry.index, dfDry["pm25b"], label="Sensit B", color="gold", alpha=0.7)
plt.plot(dfDry.index, dfDry["pm25c"], label="Sensit C", color="blue", alpha=0.7)   
plt.xlabel("Date")
plt.ylabel("PM2.5 Concentration (µg/m³)")   
plt.title("Zefan Dry Data (hourly averaged)")
plt.legend()
plt.show()
```
