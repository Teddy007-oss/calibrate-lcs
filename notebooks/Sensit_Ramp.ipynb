#SENSIT RAMP

```python
import pandas as pd

df_wet = pd.read_csv(r"C:\Users\USER\Documents\Afri-SET\Ramp Wet\ramp_wet_ref.csv",
                     parse_dates=["date"],
                     index_col="date"
                     ) 
```

```python
df_wet.describe()
```

```python
data = ["PM2.5", "pm25a", "pm25b", "pm25c"]

#DfWet is the subsetted df_wet data
dfWet = df_wet[data]

dfWet.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.plot(dfWet.index, dfWet["PM2.5"], label="Reference PM2.5", color="red")
plt.plot(dfWet.index, dfWet["pm25a"], label="Sensit A", color="green", alpha=0.7)
plt.plot(dfWet.index, dfWet["pm25b"], label="Sensit B", color="gold", alpha=0.7)
plt.plot(dfWet.index, dfWet["pm25c"], label="Sensit C", color="blue", alpha=0.7)   
plt.xlabel("Date")
plt.ylabel("PM2.5 Concentration (µg/m³)")   
plt.title("Sensit Ramp Wet Data (hourly averaged)")
plt.legend()
plt.show()
```
